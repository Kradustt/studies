```python
import requests
import json

cotacoes = requests.get ('https://economia.awesomeapi.com.br/json/all')
cotacoes_dic = cotacoes.json()
print(cotacoes_dic)
```

```python
print('Bitcoin : {}'.format(cotacoes_dic['BTC']['bid']))
print('Dólar : {}'.format(cotacoes_dic['USDT']['bid']))
print('Euro : {}'.format(cotacoes_dic['EUR']['bid']))
```

```python
#usando list comprehensions

cotacoes_dolar15d = requests.get('https://economia.awesomeapi.com.br/json/daily/USD-BRL/15')
cotacoes_dolar_dic = cotacoes_dolar15d.json()
lista_cotacao_dolar = [item['bid'] for item in cotacoes_dolar_dic]
print(lista_cotacao_dolar)
```

```python
cotacoes_btc =  requests.get('https://economia.awesomeapi.com.br/json/daily/BTC-BRL/10?start_date=20200901&end_date=20201230')
cotacoes_btc_dic = cotacoes_btc.json()
lista_cotacao_btc = [float(item['bid']) for item in cotacoes_btc_dic]
print(lista_cotacao_btc)
print(len(lista_cotacao_btc))
```

```python
import matplotlib.pyplot as plt

lista_cotacao_btc.reverse()
plt.figure(figsize=(15,3))
plt.plot(lista_cotacao_btc)

plt.show
```
