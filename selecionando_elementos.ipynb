```python
from selenium import webdriver #- quem da os comandos para abrir o navegador
from selenium.webdriver.chrome.service import Service # <- "S" Maiúsculo aqui!
from webdriver_manager.chrome import ChromeDriverManager

# 1. Cria o serviço usando a classe correta (Service com S maiúsculo)
servico = Service(ChromeDriverManager().install())

# 2. Configura as opções do Chrome
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") # Caso queira a tela cheia que vimos antes
# options.add_argument("user-data-dir=C:\User\") # Caso queira abrir já com o seu perfil do nav (chrome://version)

# 3. Abre o navegador passando TANTO o serviço QUANTO as opções
driver = webdriver.Chrome(service=servico, options=options) 

driver.get("https://www.google.com/")
```
